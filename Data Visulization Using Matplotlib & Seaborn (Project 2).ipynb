```python
# 1.SCATTER PLOT
```

```python
plt.scatter(df["Study_Hours"],df["Final_Score"])
plt.title("Performance according to study_hours")
plt.xlabel("Study_Hours")
plt.ylabel("Final_Score")
plt.show()
```

```python
# 2. BAR PLOT
```

```python
Male = (df["Gender"] == "Male")
df[Male]
```

```python
Female = (df["Gender"] == "Female")
df[Female]
```

```python
Male_Avg = df[df["Gender"] == "Male"]["Final_Score"].mean()
print(Male_Avg)
```

```python
Female_Avg = df[df["Gender"] == "Female"]["Final_Score"].mean()
print(Female_Avg)
```

```python
plt.bar(["Male","Female"],[Male_Avg,Female_Avg])
plt.xlabel("Gender")
plt.ylabel("Avg Final Score")
plt.title("Avg of Male & Female")
plt.show()
```

```python
Attendance = df["Attendance"]
print(Attendance)
```

```python
sns.scatterplot(data = df,x = "Attendance",y = "Final_Score")
plt.xlabel("Attendance")
plt.ylabel("Final_Score")
plt.title("Attendance VS final Score")
plt.show()
```

```python
Avg_of_English = df["English_Score"].mean()
print(Avg_of_English)
```

```python
Avg_of_Math = df["Math_Score"].mean()
print(Avg_of_Math)
```

```python
Avg_of_Science = df["Science_Score"].mean()
print(Avg_of_Science)
```

```python
df2 = pd.DataFrame({
    "Subject":["Maths","Science","English"],
    "Average":[Avg_of_Math,Avg_of_Science,Avg_of_English]
})
df2
```

```python
sns.barplot(data=df2,x= "Subject",y = "Average")
plt.xlabel("Subjects")
plt.ylabel("Average")
plt.show()
```
